## **NLP**

```python
import nltk #NLTK = Natural Language Toolkit
from nltk.corpus import stopwords # we take stopwords from nltk
from nltk.stem import WordNetLemmatizer #Lemmatization ka goal words ko unke base/dictionary form mein convert karna hota hai.
                                          #eg-> running  → run
import re #re = Regular Expressions , remove unwanted character like punctuation, extra space
```

```python

nltk.download('stopwords')
nltk.download('wordnet')
```

```python

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

#Data Cleaning

```python
def preprocess_text(text):
  #lowercase
  text = str(text).lower()
  #removed the punctuations
  text = re.sub(r'[^a-zA-Z\s]', '', text)

  words = text.split()
  #remove stopwords
  words = [word for word in words if word not in stop_words]

  #lemmatize
  words = [lemmatizer.lemmatize(word) for word in words]

  return " ".join(words)
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df['tags'][1]
```

```python
df = df.reset_index(drop = True)
```

```python
indices = pd.Series(df.index,index = df['title']).drop_duplicates()
indices
```

#Now we do vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english') #An n-gram is a sequence of N consecutive words/tokens in a text.
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

# **Cosine Similarity**
Cosine Similarity is a method used to measure how similar two vectors are.
TF-IDF converts movie text into vectors, and Cosine Similarity compares those vectors to find similar movies.

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n = 10):
  if title not in indices:
    return ['Movie not found']

  idx = indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

```python
recommend('Avatar',5)
```
